# Describe an image

The smallest complete mlx-vlm program: load a model, format the prompt with the
model's own chat template, generate. `num_images` tells the template how many
image placeholders to insert.

`mlx-community/Qwen3-VL-8B-Instruct-4bit` is 5.7 GB on disk and peaks at 6.82 GB,
so it runs on a 16 GB Mac.

```python
from mlx_vlm import apply_chat_template, generate, load

MODEL = "mlx-community/Qwen3-VL-8B-Instruct-4bit"
IMAGE = "../images/cats.jpg"

model, processor = load(MODEL)
```

```python
messages = [{"role": "user", "content": "Describe this image in detail."}]
prompt = apply_chat_template(processor, model.config, messages, num_images=1)

result = generate(
    model, processor, prompt, image=[IMAGE], max_tokens=150, temperature=0.0
)
print(result.text)
```

```python
print(f"{result.generation_tokens} tokens")
print(f"{result.generation_tps:.1f} tok/s decode")
print(f"{result.peak_memory:.2f} GB peak")
```

Swap `MODEL` for any supported checkpoint and `IMAGE` for a path or URL. Other
sample inputs live in `examples/images/`.
