# Several images in one prompt

Same call as a single image, with one rule: `num_images` must match the number of
images handed to `generate`. Get them out of step and the template inserts the
wrong number of placeholders, which the model misreads without raising anything.

```python
from mlx_vlm import apply_chat_template, generate, load

MODEL = "mlx-community/Qwen3-VL-8B-Instruct-4bit"
IMAGES = ["../images/demo_image1.jpg", "../images/demo_image2.png"]

model, processor = load(MODEL)
```

```python
messages = [
    {"role": "user", "content": "Compare these images. What do they have in common?"}
]
prompt = apply_chat_template(processor, model.config, messages, num_images=len(IMAGES))

result = generate(
    model, processor, prompt, image=IMAGES, max_tokens=200, temperature=0.0
)
print(result.text)
```

```python
print(f"{result.prompt_tokens} prompt tokens")
print(f"{result.generation_tps:.1f} tok/s decode")
print(f"{result.peak_memory:.2f} GB peak")
```
